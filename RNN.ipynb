```python
from keras.datasets import imdb
```

```python
(train_input, train_target), (test_input, test_target) = imdb.load_data(num_words=500)
```

```python
from sklearn.model_selection import train_test_split
train_input, val_input, train_target, val_target = train_test_split(train_input, train_target, train_size=0.2, random_state=42)
```

```python
from keras.preprocessing.sequence import pad_sequences
train_seq = pad_sequences(train_input, maxlen=100)
```

```python
val_seq = pad_sequences(val_input, maxlen=100)
```

```python
import keras
model = keras.Sequential()
model.add(keras.layers.Input(shape=(100,)))
model.add(keras.layers.Embedding(500,16))
model.add(keras.layers.GRU(8, dropout=0.2))
model.add(keras.layers.Dense(1, activation='sigmoid'))
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
checkpoint_cb = keras.callbacks.ModelCheckpoint('best-gru-model.keras', save_best_only=True)
early_stopping_cb = keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)
model.fit(train_seq, train_target, epochs=100, batch_size=64, validation_data=(val_seq, val_target), callbacks=[checkpoint_cb,early_stopping_cb])
```

```python
test_seq = pad_sequences(test_input, maxlen=100)
model.evaluate(test_seq, test_target)
```

```python
best_model = keras.models.load_model('best-gru-model.keras')
best_model.evaluate(test_seq, test_target)
```
